```python
from scipy.optimize import curve_fit
import numpy as np
```

```python
curve_fit(lambda x, a, b:  (a + (0.1 * b/5) * x ** 2)**2, np.array([3,5,7,9,11]), [0.06598, 0.06821, 0.07159, 0.07718 ,0.08536])
```

```python
curve_fit(lambda x, a, b, c:  a ** 2+ (b/5) * x ** 2 + c/25 * x ** 4, np.array([3,5,7,9,11]), [0.06598, 0.06821, 0.07159, 0.07718 ,0.08536])
```
